```python
import os
import shutil

source = r'B:\Python_projects\Skin Cancer\data_train_downsampled'
destination_NV = r'B:\Python_projects\Skin Cancer\data_final_downsam_train\NV'
destination_AKIEC = r'B:\Python_projects\Skin Cancer\data_final_downsam_train\AKIEC'
destination_BCC = r'B:\Python_projects\Skin Cancer\data_final_downsam_train\BCC'
destination_BKL = r'B:\Python_projects\Skin Cancer\data_final_downsam_train\BKL'
destination_DF = r'B:\Python_projects\Skin Cancer\data_final_downsam_train\DF'
destination_VASC = r'B:\Python_projects\Skin Cancer\data_final_downsam_train\VASC'
destination_MEL = r'B:\Python_projects\Skin Cancer\data_final_downsam_train\MEL'

# Initialize counters for each category
counters = {
    'NV': 0,
    'AKIEC': 0,
    'BCC': 0,
    'BKL': 0,
    'DF': 0,
    'VASC': 0,
    'MEL': 0
}

for i in data_downsmaple.values:
    category = i[8]
    if category in counters:
        # Increment the counter for the current category
        counters[category] += 1
        # Create a new filename with a unique identifier
        val = f"{i[0]}_{counters[category]}.jpg"
        source_path = os.path.join(source, i[0] + ".jpg")
        if category == 'NV':
            shutil.copy(source_path, os.path.join(destination_NV, val))
        elif category == 'AKIEC':
            shutil.copy(source_path, os.path.join(destination_AKIEC, val))
        elif category == 'BCC':
            shutil.copy(source_path, os.path.join(destination_BCC, val))
        elif category == 'BKL':
            shutil.copy(source_path, os.path.join(destination_BKL, val))
        elif category == 'DF':
            shutil.copy(source_path, os.path.join(destination_DF, val))
        elif category == 'VASC':
            shutil.copy(source_path, os.path.join(destination_VASC, val))
        elif category == 'MEL':
            shutil.copy(source_path, os.path.join(destination_MEL, val))
```
